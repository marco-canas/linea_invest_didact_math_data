```python
# Configuración mejorada para modelado LSTM agrícola
import tensorflow as tf
from tensorflow.keras.layers import LSTM, Dense, Input
from tensorflow.keras.models import Model
from sklearn.model_selection import TimeSeriesSplit

# Arquitectura modelo híbrido
def build_hybrid_model(input_shape):
    inputs = Input(shape=input_shape)
    x = LSTM(64, return_sequences=True, dropout=0.2)(inputs)
    x = LSTM(32, dropout=0.2)(x)
    x = Dense(16, activation='relu')(x)
    output = Dense(1, activation='sigmoid')(x)  # Probabilidad de pérdida
    
    model = Model(inputs=inputs, outputs=output)
    model.compile(optimizer='adam', loss='binary_crossentropy', 
                 metrics=['accuracy', tf.keras.metrics.AUC()])
    return model

# Validación cruzada temporal
tscv = TimeSeriesSplit(n_splits=5)
for train_idx, test_idx in tscv.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
    
    model = build_hybrid_model((X_train.shape[1], X_train.shape[2]))
    history = model.fit(X_train, y_train, epochs=50, batch_size=32,
                       validation_data=(X_test, y_test), verbose=0)
```

````python
from docx import Document
from docx.shared import Pt, Inches
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

def markdown_to_docx(markdown_text, output_filename):
    # Crear un nuevo documento
    doc = Document()
    
    # Configuración de estilos
    style = doc.styles['Normal']
    font = style.font
    font.name = 'Calibri'
    font.size = Pt(11)
    
    # Procesar el texto Markdown línea por línea
    lines = markdown_text.split('\n')
    i = 0
    while i < len(lines):
        line = lines[i].strip()
        
        # Encabezados
        if line.startswith('## '):
            add_heading(doc, line[3:], level=2)
        elif line.startswith('# '):
            add_heading(doc, line[2:], level=1)
        
        # Tablas
        elif '|' in line and '---' not in line:
            table_data = []
            # Obtener todas las filas de la tabla
            while i < len(lines) and '|' in lines[i]:
                if '---' not in lines[i]:  # Saltar la línea de separación
                    row = [cell.strip() for cell in lines[i].split('|') if cell.strip()]
                    table_data.append(row)
                i += 1
            i -= 1  # Retroceder un paso
            
            # Crear la tabla en el documento
            if table_data:
                table = doc.add_table(rows=len(table_data), cols=len(table_data[0]))
                table.style = 'Table Grid'
                
                for row_idx, row in enumerate(table_data):
                    for col_idx, cell in enumerate(row):
                        table.cell(row_idx, col_idx).text = cell
                        # Centrar texto en las celdas
                        paragraph = table.cell(row_idx, col_idx).paragraphs[0]
                        paragraph.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
        
        # Listas de items
        elif line.startswith('- '):
            p = doc.add_paragraph(style='List Bullet')
            p.add_run(line[2:])
        
        # Texto normal
        elif line:
            doc.add_paragraph(line)
        
        i += 1
    
    # Guardar el documento
    doc.save(output_filename)
    print(f"Documento guardado como {output_filename}")

def add_heading(doc, text, level):
    heading = doc.add_heading(text, level=level)
    # Configurar fuente para los encabezados
    for run in heading.runs:
        run.font.name = 'Calibri'
        run.font.size = Pt(14 if level == 1 else 12)
        run.font.bold = True

# Tu texto Markdown
markdown_text = """
# Cronograma Ajustado para 2 Años y Entregables para el Proyecto CODI

## Cronograma Extendido (24 meses)

| Actividad | Trim 1-2 (M1-6) | Trim 3-4 (M7-12) | Trim 5-6 (M13-18) | Trim 7-8 (M19-24) |
|-----------|-----------------|------------------|-------------------|-------------------|
| Recolección y procesamiento de datos | ✓ | ✓ | | |
| Desarrollo de modelos predictivos | ✓ | ✓ | ✓ | |
| Implementación de estaciones meteorológicas | | ✓ | ✓ | |
| Validación en campo | | | ✓ | ✓ |
| Desarrollo de interfaz y aplicación | | ✓ | ✓ | ✓ |
| Capacitación a agricultores | | | ✓ | ✓ |
| Redacción de informes y publicaciones | | | ✓ | ✓ |
| Transferencia tecnológica | | | | ✓ |

## Entregables Clave por Fase:

**Fase 1: Diagnóstico y diseño (Meses 1-6)**
- Informe de caracterización climática y agrícola del Bajo Cauca
- Diseño conceptual de la plataforma predictiva
- Primer prototipo de modelo predictivo básico (ARIMA)

**Fase 2: Desarrollo técnico (Meses 7-12)**
- Manual de usuario preliminar
- Versión beta de la aplicación con módulo de predicción
- Informe técnico de modelos implementados (ARIMA + LSTM)
- 2 estaciones meteorológicas piloto instaladas

**Fase 3: Validación (Meses 13-18)**
- Reporte de validación con agricultores (50 casos)
- Versión estable de la aplicación móvil
- Manual de usuario finalizado
- Paper científico enviado a revista indexada

**Fase 4: Implementación (Meses 19-24)**
- Informe final con análisis de impacto
- Plataforma en funcionamiento con al menos 100 usuarios
- Plan de sostenibilidad y escalamiento
- Talleres de transferencia tecnológica completados
- Documentación técnica completa en repositorio GitHub
"""

# Convertir y guardar
markdown_to_docx(markdown_text, "Cronograma_Proyecto_CODI.docx")
````

````python
from docx import Document
from docx.shared import Pt, Inches, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

def set_cell_background(cell, color):
    """Fondo de celda con color (en formato HEX)"""
    shading_elm = OxmlElement('w:shd')
    shading_elm.set(qn('w:fill'), color)
    cell._tc.get_or_add_tcPr().append(shading_elm)

def create_budget_docx(output_filename):
    doc = Document()
    
    # Configuración general
    style = doc.styles['Normal']
    font = style.font
    font.name = 'Arial'
    font.size = Pt(10)
    
    # Título del documento
    title = doc.add_heading('PRESUPUESTO AJUSTADO (≤ $100 MILLONES COP)', level=1)
    title.alignment = WD_ALIGN_PARAGRAPH.CENTER
    for run in title.runs:
        run.font.color.rgb = RGBColor(0, 32, 96)  # Azul institucional
        run.font.bold = True
    
    # Subtítulo
    subtitle = doc.add_heading('Proyecto: Aplicación de Análisis Predictivo para Gestión de Riesgos Climáticos', level=2)
    subtitle.alignment = WD_ALIGN_PARAGRAPH.CENTER
    
    # Crear tabla de presupuesto
    table = doc.add_table(rows=1, cols=4)
    table.style = 'Table Grid'
    
    # Encabezados de tabla
    hdr_cells = table.rows[0].cells
    headers = ['Rubro', 'Detalle', 'Valor (COP)', 'Ajustes Realizados']
    for i, header in enumerate(headers):
        hdr_cells[i].text = header
        hdr_cells[i].paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.CENTER
        set_cell_background(hdr_cells[i], 'D9E1F2')  # Azul claro
        for run in hdr_cells[i].paragraphs[0].runs:
            run.font.bold = True
    
    # Datos del presupuesto
    budget_data = [
        ["1. Equipos", 
         "2 Estaciones meteorológicas básicas + sensores de humedad del suelo + 2 tablets", 
         "$24,000,000", 
         "Reducción a 2 estaciones y sensores críticos"],
        
        ["2. Trabajo de campo", 
         "Logística (18 meses) + 2 técnicos medio tiempo", 
         "$24,000,000", 
         "Personal local y transporte optimizado"],
        
        ["3. Desarrollo de software", 
         "Ingeniero (12 meses) + Desarrollador (6 meses) + Cloud", 
         "$36,000,000", 
         "Enfoque en MVP con herramientas open-source"],
        
        ["4. Gestión del conocimiento", 
         "1 artículo Q4 + 3 talleres + Plataforma web + Video", 
         "$12,000,000", 
         "Publicación priorizada y materiales básicos"],
        
        ["5. Imprevistos (10%)", 
         "Fondo para contingencias", 
         "$4,000,000", 
         "Reducción controlada"],
        
        ["TOTAL", 
         "", 
         "$100,000,000", 
         ""]
    ]
    
    # Añadir filas a la tabla
    for row_data in budget_data:
        row_cells = table.add_row().cells
        for i, cell_value in enumerate(row_data):
            row_cells[i].text = cell_value
            if i == 2:  # Alinear valores a la derecha
                row_cells[i].paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.RIGHT
    
    # Formato especial para la fila TOTAL
    total_row = table.rows[-1]
    for cell in total_row.cells[:3]:
        set_cell_background(cell, 'F8CBAD')  # Naranja claro
        for paragraph in cell.paragraphs:
            for run in paragraph.runs:
                run.font.bold = True
    
    # Notas al pie
    doc.add_paragraph().add_run("Notas:").bold = True
    notes = [
        "1. Costos basados en cotizaciones locales de 2024",
        "2. Incluye alianzas estratégicas con UdeA para reducir costos",
        "3. MVP: Minimum Viable Product (Producto Mínimo Viable)",
        "4. Precios en pesos colombianos (COP)"
    ]
    for note in notes:
        doc.add_paragraph(note, style='List Bullet')
    
    # Guardar documento
    doc.save(output_filename)
    print(f"Documento generado: {output_filename}")

# Ejecutar la función
create_budget_docx("Presupuesto_Proyecto_CODI.docx")
````
