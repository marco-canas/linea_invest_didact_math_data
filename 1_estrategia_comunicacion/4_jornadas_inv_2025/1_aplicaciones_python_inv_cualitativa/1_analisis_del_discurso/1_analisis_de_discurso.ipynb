````python
import pandas as pd
import os
from datetime import datetime

def registrar_asistente(nombre, correo, institucion, archivo_excel):
    """
    Registra un asistente en el archivo Excel de la master class.
    
    Parámetros:
    - nombre (str): Nombre completo del asistente
    - correo (str): Correo electrónico del asistente
    - institucion (str): Institución de procedencia
    - archivo_excel (str): Ruta del archivo Excel
    
    Si el archivo no existe, lo crea con las columnas necesarias.
    """
    
    # Verificar si el archivo ya existe
    if os.path.exists(archivo_excel):
        df = pd.read_excel(archivo_excel)
    else:
        # Crear un nuevo DataFrame si no existe
        df = pd.DataFrame(columns=["Nombre", "Correo", "Institución", "Fecha_Registro"])
    
    # Crear registro nuevo
    nuevo = {
        "Nombre": nombre,
        "Correo": correo,
        "Institución": institucion,
        "Fecha_Registro": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    
    # Agregar al DataFrame
    df = pd.concat([df, pd.DataFrame([nuevo])], ignore_index=True)
    
    # Guardar de nuevo en el Excel
    df.to_excel(archivo_excel, index=False)
    
    print(f"Asistente {nombre} registrado con éxito en {archivo_excel}.")
````

# 📚 Introducción a las herramientas de Python para textos (30 min)

La investigación cualitativa en Administración de Empresas cada vez depende más de herramientas tecnológicas que permitan **organizar, procesar y visualizar grandes cantidades de información textual**: entrevistas, etnografías, encuestas abiertas, reportes de gerencia, comentarios de clientes, entre otros.

En este taller nos apoyaremos en un conjunto de librerías de Python que se han convertido en **estándares para el análisis de textos** en las ciencias sociales y económicas.

---

## 🔹 Librerías básicas

### 1. `pandas` → manejo de datos textuales en tablas

* **¿Qué hace?** Permite leer, organizar y manipular datos en formato tabular (Excel, CSV).
* **Ejemplo aplicado:** almacenar en un *DataFrame* las transcripciones de entrevistas a empresarios o empleados, y poder filtrar, ordenar y seleccionar fragmentos de interés.

```python
import pandas as pd
data = pd.DataFrame({"Entrevistado": ["Gerente", "Empleado"], 
                     "Texto": ["La productividad es clave", "Nos preocupa el estrés"]})
print(data)
```

---

### 2. `nltk` → procesamiento básico del lenguaje

* **¿Qué hace?** Permite trabajar con el lenguaje natural: separar un texto en palabras (*tokenización*), eliminar palabras vacías (*stopwords*), y reducir palabras a su raíz (*stemming* o *lematización*).
* **Ejemplo aplicado:** preparar los discursos de gerentes y empleados para comparar qué términos son más frecuentes.

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')

texto = "La comunicación en la empresa es fundamental para los resultados."
tokens = word_tokenize(texto.lower())
print(tokens)
```

---

### 3. `wordcloud` → nubes de palabras

* **¿Qué hace?** Genera representaciones visuales donde el tamaño de cada palabra depende de su frecuencia.
* **Ejemplo aplicado:** comparar el vocabulario de gerentes vs. empleados en una nube de palabras.

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

texto = "estrategia resultados innovación clientes productividad"
wc = WordCloud(background_color="white").generate(texto)
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```

---

### 4. `matplotlib` → visualización de resultados

* **¿Qué hace?** Es la librería más usada para gráficos en Python. Nos permite representar patrones de uso de palabras o comparaciones entre textos.
* **Ejemplo aplicado:** graficar la frecuencia de palabras clave en entrevistas a distintos actores de una organización.

```python
import matplotlib.pyplot as plt

palabras = ["estrategia", "clientes", "costos", "innovación"]
frecuencia = [10, 7, 4, 6]

plt.bar(palabras, frecuencia)
plt.title("Frecuencia de términos en entrevistas")
plt.show()
```

---

### 5. `scikit-learn` → análisis más avanzado

* **¿Qué hace?** Ofrece herramientas de minería de texto como el **TF-IDF** (medida de importancia de palabras en documentos) y técnicas básicas de clustering o clasificación.
* **Ejemplo aplicado:** identificar qué temas diferencian los discursos de distintos grupos de entrevistados (ej. gerentes vs. empleados).

```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "La productividad y los clientes son prioridad",
    "Nos preocupa el clima organizacional y el estrés"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

````python
import pandas as pd
import nltk
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Texto simulado de discursos de gerentes
textos = [
    "Nuestro liderazgo se centra en la innovación y la confianza del cliente",
    "La crisis nos enseñó a valorar el trabajo en equipo y la resiliencia",
    "El futuro de la administración está en la transformación digital"
]

# Convertir a DataFrame
df = pd.DataFrame(textos, columns=["Discurso"])
df

````

### 3. **Procesamiento básico del discurso** (40 min)

* **Tokenización** (separar palabras).
* **Eliminación de stopwords** (palabras vacías).
* **Lematización** (normalizar palabras).



👉 **Ejemplo práctico**:


````python
import nltk
import pandas as pd

# Descargar los recursos necesarios
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.corpus import stopwords

# Definir stopwords en español
stop_words = set(stopwords.words('spanish'))

# Supongamos que ya tienes tu DataFrame df con la columna "Discurso"
# Ejemplo de prueba:
# df = pd.DataFrame({"Discurso": ["La productividad es esencial en la empresa", "Nos preocupa el clima organizacional"]})

# Tokenización + limpieza
df["Tokens"] = df["Discurso"].apply(
    lambda x: [
        word.lower() 
        for word in nltk.word_tokenize(x, language="spanish") 
        if word.isalpha() and word.lower() not in stop_words
    ]
)

df

````

````python
df
````


---

### 4. **Visualización del discurso** (30 min)

* Construcción de **nube de palabras** para ver los conceptos más recurrentes.

👉 **Ejemplo práctico**:


````python
all_words = " ".join([" ".join(tokens) for tokens in df["Tokens"]])
wordcloud = WordCloud(width=800, height=400, background_color="white").generate(all_words)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()

````


---

### 5. **Análisis más profundo: TF-IDF y clusters** (30 min)

* Explicación de la técnica **TF-IDF** para identificar las palabras más relevantes en cada discurso.
* Uso de **clustering (KMeans)** para agrupar discursos con similitudes.

👉 **Ejemplo práctico simplificado**:


````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

vectorizer = TfidfVectorizer(stop_words=stop_words)
X = vectorizer.fit_transform(df["Discurso"])

kmeans = KMeans(n_clusters=2, random_state=42)
df["Cluster"] = kmeans.fit_predict(X)

df

````
