```python
import pandas as pd
import os
from datetime import datetime

def registrar_asistente(nombre, correo, institucion, archivo_excel):
    """
    Registra un asistente en el archivo Excel de la master class.
    
    Parámetros:
    - nombre (str): Nombre completo del asistente
    - correo (str): Correo electrónico del asistente
    - institucion (str): Institución de procedencia
    - archivo_excel (str): Ruta del archivo Excel
    
    Si el archivo no existe, lo crea con las columnas necesarias.
    """
    
    # Verificar si el archivo ya existe
    if os.path.exists(archivo_excel):
        df = pd.read_excel(archivo_excel)
    else:
        # Crear un nuevo DataFrame si no existe
        df = pd.DataFrame(columns=["Nombre", "Correo", "Institución", "Fecha_Registro"])
    
    # Crear registro nuevo
    nuevo = {
        "Nombre": nombre,
        "Correo": correo,
        "Institución": institucion,
        "Fecha_Registro": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    
    # Agregar al DataFrame
    df = pd.concat([df, pd.DataFrame([nuevo])], ignore_index=True)
    
    # Guardar de nuevo en el Excel
    df.to_excel(archivo_excel, index=False)
    
    print(f"Asistente {nombre} registrado con éxito en {archivo_excel}.")
```

### 3. `wordcloud` → nubes de palabras

* **¿Qué hace?** Genera representaciones visuales donde el tamaño de cada palabra depende de su frecuencia.
* **Ejemplo aplicado:** comparar el vocabulario de gerentes vs. empleados en una nube de palabras.


```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Agregar stopwords en español
stopwords_es = set(STOPWORDS)
stopwords_es.update([
    "la", "el", "los", "las", "de", "del", "y", "a", "en", "para", "por", "con",
    "se", "una", "un", "al", "o", "que", "su", "sus", "como", "es", "son", "e",
    "lo", "más", "ya", "también", "ha", "han", "ha", "sin", "desde", "sobre",
    "entre", "hasta", "cuando", "porque", "si", "donde", "cuya", "cuyo"
])

texto = """La administración empresarial combina conocimientos de economía, 
finanzas, marketing, operaciones y recursos humanos para construir organizaciones 
sólidas y adaptativas. A través de la planificación estratégica se definen metas 
claras y se asignan recursos para alcanzarlas; la organización traduce la estrategia 
en estructuras, responsabilidades y procesos eficientes. La dirección motiva y coordina equipos, 
desarrolla habilidades y promueve una cultura orientada al cliente y al aprendizaje. 
El control monitorea resultados, facilita la retroalimentación y ajusta acciones 
para mejorar el rendimiento. En un contexto globalizado, la competitividad exige 
innovación constante, investigación de mercado, agilidad para responder a cambios 
tecnológicos y políticas de sostenibilidad ambiental y social. La digitalización 
permite automatizar procesos, analizar grandes volúmenes de datos y personalizar 
la experiencia del cliente. La gestión financiera asegura liquidez, rentabilidad y 
sostenibilidad, mediante presupuestos, control de costos y análisis de inversiones. 
Asimismo, el gobierno corporativo y la ética empresarial protegen la reputación y 
generan confianza con stakeholders. Finalmente, la colaboración interdepartamental y 
la formación continua son esenciales para mantener ventajas estratégicas y crear valor
compartido para la comunidad.
"""

# Generar la nube de palabras
wc = WordCloud(
    background_color="white",
    stopwords=stopwords_es,
    width=800,
    height=400,
    collocations=False
).generate(texto)

# Mostrar la nube
plt.figure(figsize=(10,5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```

# 3. **Procesamiento básico del discurso** (40 min)

* **Tokenización** (separar palabras).
* **Eliminación de stopwords** (palabras vacías).
* **Lematización** (normalizar palabras).



## **Ejemplo práctico**:


```python
import nltk
import pandas as pd

# Descargar los recursos necesarios
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.corpus import stopwords

# Definir stopwords en español
stop_words = set(stopwords.words('spanish'))

# Supongamos que ya tienes tu DataFrame df con la columna "Discurso"
# Ejemplo de prueba:
# df = pd.DataFrame({"Discurso": ["La productividad es esencial en la empresa", "Nos preocupa el clima organizacional"]})

# Tokenización + limpieza
df["Tokens"] = df["Discurso"].apply(
    lambda x: [
        word.lower() 
        for word in nltk.word_tokenize(x, language="spanish") 
        if word.isalpha() and word.lower() not in stop_words
    ]
)

df

```


# 5. **Análisis más profundo: TF-IDF y clusters** (30 min)

* Explicación de la técnica **TF-IDF** para identificar las palabras más relevantes en cada discurso.
* Uso de **clustering (KMeans)** para agrupar discursos con similitudes.



## **Ejemplo práctico simplificado**:


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Aseguramos que 'stop_words' sea una lista (no un set)
stop_words = [
    "a", "al", "algo", "algunos", "como", "con", "contra", "de", "del", "desde", "donde",
    "el", "ella", "ellas", "ellos", "en", "entre", "era", "eran", "es", "esa", "ese",
    "eso", "esta", "estas", "este", "estos", "ha", "han", "hay", "la", "las", "le", "les",
    "lo", "los", "más", "me", "mi", "mis", "mucho", "muy", "no", "nos", "nuestra", "nuestro",
    "o", "os", "para", "pero", "por", "que", "quien", "se", "sin", "sobre", "su", "sus",
    "te", "ti", "tus", "tu", "un", "una", "uno", "y", "ya"
]

# Convertimos los NaN en cadenas vacías
df["Discurso"] = df["Discurso"].fillna("")

# Creamos el vectorizador con las stopwords personalizadas
vectorizer = TfidfVectorizer(stop_words=stop_words)

# ✅ Transformamos los textos a la matriz TF-IDF
X = vectorizer.fit_transform(df["Discurso"])

# ✅ Creamos el modelo KMeans
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

# ✅ Etiquetamos los clusters en el DataFrame
df["Cluster"] = kmeans.fit_predict(X)

df

```
