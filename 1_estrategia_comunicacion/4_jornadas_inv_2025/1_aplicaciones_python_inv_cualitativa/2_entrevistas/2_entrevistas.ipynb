````python
import qrcode
from IPython.display import display

def generar_qr_notebook(url):
    """
    Genera un código QR para una URL y lo muestra en un cuaderno Jupyter.
    
    Parámetros:
    url (str): La URL para la cual se generará el código QR.
    
    Retorna:
    None (pero muestra el código QR en el notebook)
    """
    # Crear el objeto QR
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    
    # Añadir la URL al código QR
    qr.add_data(url)
    qr.make(fit=True)
    
    # Crear la imagen del QR
    img = qr.make_image(fill_color="black", back_color="white")
    
    # Mostrar la imagen en el notebook
    display(img)

````

# 1. Entrevistas


Son una técnica de recolección de información cualitativa que busca comprender las percepciones, experiencias, ideas o significados que las personas atribuyen a un fenómeno.
En investigación educativa, social o administrativa, se utilizan para obtener **discursos ricos y detallados** sobre un tema.



### Tipos comunes

* **Estructuradas:** preguntas fijas, iguales para todos los entrevistados.
* **Semiestructuradas:** combinan preguntas guía con la posibilidad de explorar respuestas emergentes (las más usadas en investigación cualitativa).
* **No estructuradas:** el diálogo fluye libremente; útiles en etapas exploratorias.



### Objetivo

Capturar **la voz del participante** para luego analizar su discurso y **extraer significados**, categorías y patrones.

# 2. Codificación


La **codificación** es el proceso de **asignar etiquetas o códigos** a fragmentos de texto (palabras, frases, párrafos) que representan una idea o categoría relevante para la investigación.



En otras palabras, **traducimos el lenguaje natural a datos analíticos**.
Ejemplo:

> Fragmento: “Me gusta trabajar con Python porque facilita el análisis de datos.”
> Código: `motivación_aprendizaje`, `uso_Python`, `análisis_datos`



### En Python

* Se pueden **automatizar partes del proceso** con herramientas de análisis de texto, por ejemplo:

  * `nltk` o `spaCy` para tokenizar y limpiar el texto.
  * `TfidfVectorizer` o `CountVectorizer` (de `scikit-learn`) para representar el discurso como una matriz numérica.
  * Algoritmos de **clustering (como KMeans)** para agrupar respuestas similares.

Esto no reemplaza la interpretación humana, pero **ayuda a organizar, reducir y visualizar** grandes volúmenes de texto.

# 3. Patrones temáticos

1. Familiarización con los datos.
2. Generación de códigos iniciales.
3. Búsqueda de temas.
4. Revisión de temas.
5. Definición y nombramiento de temas.
6. Elaboración del informe.

* **Frecuencia de palabras o nubes de palabras** (`wordcloud`, `Counter` de `collections`).
* **Modelos de tópicos** como *Latent Dirichlet Allocation (LDA)* para descubrir temas latentes.
* **Agrupamientos** (*clustering*) por similitud semántica con `KMeans` o `AgglomerativeClustering`.
* **Visualizaciones** con `matplotlib`, `seaborn` o `plotly` para mostrar relaciones entre temas y categorías.

# 1. Exploración y limpieza del texto



El primer paso en el análisis textual es **preparar el corpus**, es decir, limpiar y normalizar las entrevistas.
Python permite automatizar esta tarea fácilmente con librerías como:

* **`pandas`** → para almacenar y manipular entrevistas en un DataFrame.
* **`re`** (expresiones regulares) → para eliminar símbolos o caracteres innecesarios.
* **`nltk`** o **`spaCy`** → para tokenizar (dividir el texto en palabras), convertir a minúsculas, eliminar “stopwords” y lematizar.



**Ejemplo:**


````python

import pandas as pd
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

# Cargar entrevistas
df = pd.read_csv("entrevistas.csv")

# Limpieza básica
stop_words = stopwords.words('spanish')
df["texto_limpio"] = df["entrevista"].str.lower()
df["texto_limpio"] = df["texto_limpio"].apply(
    lambda x: " ".join([w for w in x.split() if w not in stop_words])
)

````

# **4. Limpieza y preprocesamiento de texto**

* Convertir a minúsculas
* Eliminar signos de puntuación

````python
import nltk
import re
from nltk.corpus import stopwords

nltk.download("stopwords")
stop_words = set(stopwords.words("spanish"))

def limpiar_texto(texto):
    texto = texto.lower()
    texto = re.sub(r"[^a-záéíóúñ\s]", "", texto)
    palabras = [p for p in texto.split() if p not in stop_words]
    return " ".join(palabras)

df["Respuesta_limpia"] = df["Respuesta"].apply(limpiar_texto)
df

````

# from sklearn.feature_extraction.text import TfidfVectorizer  



El módulo `TfidfVectorizer` de **scikit-learn** (`from sklearn.feature_extraction.text import TfidfVectorizer`) es una herramienta muy útil para transformar texto en datos numéricos que pueden ser analizados estadística o computacionalmente. 

Aunque proviene del campo del **procesamiento del lenguaje natural (NLP)**, su uso se ha extendido a la **investigación cualitativa** —por ejemplo, para el análisis de entrevistas, grupos focales o diarios de campo—, ya que permite **cuantificar patrones de lenguaje y temas emergentes**.


# 1. Qué es el `TfidfVectorizer`

`TfidfVectorizer` significa **Term Frequency–Inverse Document Frequency Vectorizer**.
Es una técnica de representación de texto que mide **la importancia relativa de una palabra en un documento dentro de un conjunto de documentos**.



La idea es simple:

* **TF (Term Frequency)**: mide cuántas veces aparece una palabra en un documento.
  $$
  TF(t, d) = \frac{\text{Número de veces que aparece el término } t \text{ en el documento } d}{\text{Número total de términos en } d}
  $$



* **IDF (Inverse Document Frequency)**: mide cuán exclusiva es una palabra en todo el conjunto de documentos.
  $$
  IDF(t) = \log\left(\frac{N}{1 + n_t}\right)
  $$
  donde $N$ es el número total de documentos y $n_t$ es el número de documentos que contienen el término $t$.



* **TF-IDF** es el producto de ambos:
  $$
  TFIDF(t, d) = TF(t, d) \times IDF(t)
  $$
  Cuanto más frecuente sea una palabra en un documento y menos frecuente en el resto, **mayor será su peso TF-IDF**.




# 2. Aplicación en análisis de entrevistas cualitativas

En **investigación cualitativa**, este vectorizador puede servir como puente entre los enfoques interpretativos y los cuantitativos. Algunas aplicaciones concretas son:



## a) Identificación de temas y patrones de lenguaje

Si tienes varias entrevistas transcritas, `TfidfVectorizer` te permite cuantificar la frecuencia relativa de palabras clave por entrevista.
Por ejemplo:


````python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Ejemplo de corpus de entrevistas
corpus = [
    "Los estudiantes sienten ansiedad frente a los exámenes.",
    "La motivación depende del acompañamiento del docente.",
    "Los estudiantes muestran mayor interés cuando hay actividades prácticas."
]

vectorizer = TfidfVectorizer(stop_words='spanish')
X = vectorizer.fit_transform(corpus)

df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
df

````


Esto genera una **matriz TF-IDF** donde las filas representan entrevistas y las columnas palabras relevantes.
Podrás observar qué términos son más distintivos de cada discurso.

## b) Agrupación o segmentación de discursos (clustering)

Ejemplo:


````python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42)
df["cluster"] = kmeans.fit_predict(X)

````


Esto puede revelar, por ejemplo, que **un grupo de entrevistados habla más de “motivación y acompañamiento”** y otro más de **“ansiedad y evaluación”**.

# **Análisis de entrevistas con TF-IDF y K-Means en investigación cualitativa**


````python
# ============================================
# Análisis de entrevistas cualitativas con TF-IDF y K-Means
# Tema: Emociones de los estudiantes frente a la evaluación
# Autor: Marco Julio Cañas
# ============================================

# 1. Importar librerías
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# 2. Corpus de entrevistas simuladas
entrevistas = [
    "Me siento muy nervioso antes de los exámenes, temo no recordar lo que estudié.",
    "Cuando el profesor explica bien, me siento más tranquilo durante la evaluación.",
    "La ansiedad me bloquea y no puedo concentrarme al responder.",
    "Siento motivación cuando los exámenes son prácticos y relacionados con la vida real.",
    "Me frustra cuando no entiendo lo que me preguntan en el examen.",
    "Disfruto los exámenes cuando puedo aplicar lo aprendido en proyectos.",
    "Tengo miedo a equivocarme y que el profesor piense que no estudié.",
    "Me gusta cuando el profesor nos da retroalimentación después de la prueba.",
    "Pierdo el interés cuando las evaluaciones son solo de memoria.",
    "Me siento más confiado cuando estudio con mis compañeros."
]

df = pd.DataFrame({"Entrevista": entrevistas})
df

````


# 3. Vectorización con **TF-IDF**


````python

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.corpus import stopwords

# Descargar stopwords si no las tienes
import nltk
nltk.download('stopwords')

# Cargar stopwords en español
spanish_stopwords = stopwords.words('spanish')

# Crear vectorizador TF-IDF con stopwords en español
vectorizer = TfidfVectorizer(stop_words=spanish_stopwords)

# Transformar el texto a una matriz numérica
X = vectorizer.fit_transform(df["Entrevista"].astype(str))  # Asegura que sean strings

# Convertir a DataFrame para observar las palabras y sus pesos
tfidf_df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
tfidf_df.head()


````


**Interpretación:**
Cada fila representa una entrevista y cada columna una palabra relevante.
Los valores indican **qué tan distintiva** es una palabra en esa entrevista.

---



# 4. Agrupar entrevistas según similitud semántica con **K-Means**


````python

# Aplicar K-Means (2 grupos)
kmeans = KMeans(n_clusters=2, random_state=42)
df["Cluster"] = kmeans.fit_predict(X)

df

````


**Interpretación:**
El algoritmo ha agrupado las entrevistas en **dos grupos semánticos**.
Por ejemplo:

* Cluster 0 → entrevistas centradas en **ansiedad, nervios y miedo**.
* Cluster 1 → entrevistas centradas en **motivación, comprensión y disfrute**.





## 5. Analizar las palabras más representativas de cada cluster


````python

# Identificar palabras representativas por cluster
import numpy as np

palabras = vectorizer.get_feature_names_out()
clusters = kmeans.labels_

# Promedio de TF-IDF por cluster
mean_tfidf = X.toarray().T.dot(pd.get_dummies(clusters)) / np.bincount(clusters)

for i in range(kmeans.n_clusters):
    top_idx = mean_tfidf[:, i].argsort()[::-1][:10]
    top_words = [palabras[j] for j in top_idx]
    print(f"\n Cluster {i} - Palabras clave:")
    print(", ".join(top_words))

````


# 6. Visualización: Nubes de palabras por grupo


````python

# Crear nubes de palabras por cluster
for i in range(kmeans.n_clusters):
    texto_cluster = " ".join(df[df["Cluster"] == i]["Entrevista"])
    nube = WordCloud(width=800, height=400, background_color='white', \
        colormap='plasma').generate(texto_cluster)
    plt.figure(figsize=(8,4))
    plt.imshow(nube, interpolation='bilinear')
    plt.axis('off')
    plt.title(f"Nube de palabras - Cluster {i}")
    plt.show()

````


**Interpretación visual:**
Cada nube muestra los términos más frecuentes (y distintivos) de las entrevistas agrupadas.
Sirve como apoyo para **codificar temas** en tu análisis cualitativo.

# 1.2. Fundamento matemático del modelo TF-IDF

El vectorizador **TF-IDF (Term Frequency – Inverse Document Frequency)** transforma textos en números midiendo la **importancia relativa** de cada palabra en un conjunto de documentos.

$$
TFIDF(t, d) = TF(t, d) \times IDF(t)
$$

Donde:

$$
TF(t, d) = \frac{\text{Número de veces que aparece el término } t \text{ en el documento } d}{\text{Número total de términos en } d}
$$



$$
IDF(t) = \log\left(\frac{N}{1 + n_t}\right)
$$

* $t$: término o palabra
* $d$: documento (entrevista)
* $N$: número total de documentos
* $n_t$: número de documentos donde aparece el término $t$



El resultado es un valor alto cuando la palabra es **frecuente en un documento pero rara en el resto**, lo que indica **especificidad temática**.

## 2.1. Importar librerías


````python

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import numpy as np

````

## 2.2. Corpus de entrevistas simuladas


````python
# Corpus representativo de entrevistas sobre emociones en la evaluación
entrevistas = [
    "Me siento muy nervioso antes de los exámenes, temo no recordar lo que estudié.",
    "Cuando el profesor explica bien, me siento más tranquilo durante la evaluación.",
    "La ansiedad me bloquea y no puedo concentrarme al responder.",
    "Siento motivación cuando los exámenes son prácticos y relacionados con la vida real.",
    "Me frustra cuando no entiendo lo que me preguntan en el examen.",
    "Disfruto los exámenes cuando puedo aplicar lo aprendido en proyectos.",
    "Tengo miedo a equivocarme y que el profesor piense que no estudié.",
    "Me gusta cuando el profesor nos da retroalimentación después de la prueba.",
    "Pierdo el interés cuando las evaluaciones son solo de memoria.",
    "Me siento más confiado cuando estudio con mis compañeros."
]

df = pd.DataFrame({"Entrevista": entrevistas})
df

````

# 2.3. Representación TF-IDF


````python
vectorizer = TfidfVectorizer(stop_words='spanish')
X = vectorizer.fit_transform(df["Entrevista"])

# Mostrar matriz TF-IDF
tfidf_df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
tfidf_df.head()
````


**Interpretación:**
Cada valor refleja la **relevancia** de una palabra en una entrevista.
Palabras comunes (como “profesor” o “evaluación”) pueden tener valores bajos si aparecen en muchos discursos, mientras que palabras distintivas (como “nervioso” o “proyectos”) tendrán valores altos.





# 2.4. Agrupación semántica con K-Means


````python
kmeans = KMeans(n_clusters=2, random_state=42)
df["Cluster"] = kmeans.fit_predict(X)
df

````


**Análisis inicial:**

* `Cluster 0`: puede agrupar discursos con **ansiedad, miedo y nerviosismo**.
* `Cluster 1`: puede agrupar discursos con **motivación, confianza y disfrute**.





# 2.5. Palabras más representativas por grupo


````python
palabras = vectorizer.get_feature_names_out()
clusters = kmeans.labels_

mean_tfidf = X.toarray().T.dot(pd.get_dummies(clusters)) / np.bincount(clusters)

for i in range(kmeans.n_clusters):
    top_idx = mean_tfidf[:, i].argsort()[::-1][:10]
    top_words = [palabras[j] for j in top_idx]
    print(f"\n Cluster {i} - Palabras clave:")
    print(", ".join(top_words))

````


# 2.6. Visualización con nubes de palabras


````python
for i in range(kmeans.n_clusters):
    texto_cluster = " ".join(df[df["Cluster"] == i]["Entrevista"])
    nube = WordCloud(width=800, height=400, background_color='white', colormap='plasma').generate(texto_cluster)
    plt.figure(figsize=(8,4))
    plt.imshow(nube, interpolation='bilinear')
    plt.axis('off')
    plt.title(f"Nube de palabras - Cluster {i}")
    plt.savefig(r"C:\Users\marco\Downloads\nube_de_palabras_cluster.png")
    plt.show()

````

# Apéndice sobre kmeans  




Este fragmento de código aplica **K-means clustering** para **agrupar respuestas de texto** (por ejemplo, de entrevistas, cuestionarios o encuestas) de acuerdo con su **similitud semántica**, medida a través de **vectores TF-IDF**.


Vamos a desglosar el proceso paso a paso:

---



### 🧩 1. Importación de librerías


````python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

````


* **`TfidfVectorizer`** convierte el texto en una **matriz numérica** donde cada columna representa una palabra y cada fila un documento (en este caso, una respuesta).

  * TF-IDF significa *Term Frequency – Inverse Document Frequency*.
  * Evalúa qué tan importante es una palabra dentro de un documento respecto al conjunto total de documentos.
* **`KMeans`** es un algoritmo de aprendizaje no supervisado que **agrupa automáticamente datos en K grupos (clusters)**.

---



# 2. Representación vectorial del texto


````python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["Respuesta_limpia"])

````


* Aquí, `df["Respuesta_limpia"]` contiene el texto procesado (sin stopwords, minúsculas, sin puntuación, etc.).
* `fit_transform()`:

  * **fit:** aprende el vocabulario (las palabras únicas).
  * **transform:** convierte cada respuesta en un **vector TF-IDF**.
* El resultado, `X`, es una **matriz dispersa** de tamaño `(n_respuestas, n_palabras)`.

*Cada respuesta se convierte en un punto en un espacio de alta dimensión, donde las coordenadas reflejan la importancia de las palabras.*

---



# 3. Creación del modelo K-means


````python

kmeans = KMeans(n_clusters=2, random_state=42)

````


* Se crea un modelo K-means con **2 clusters**.
* `random_state=42` asegura reproducibilidad.
* El algoritmo K-means intentará **dividir los datos en 2 grupos** según la similitud entre los vectores TF-IDF.

---



## 4. Entrenamiento del modelo


````python

kmeans.fit(X)

````


Aquí ocurre el proceso de **clustering**, que sigue estos pasos:

1. **Inicialización**: elige aleatoriamente 2 centroides (uno para cada cluster).
2. **Asignación**: cada respuesta se asigna al cluster cuyo centroide esté más cercano (según la distancia euclidiana en el espacio TF-IDF).
3. **Actualización**: se recalculan los centroides como el promedio de los puntos de cada cluster.
4. **Iteración**: se repiten los pasos 2 y 3 hasta que las asignaciones ya no cambian significativamente o se alcanza el número máximo de iteraciones.



El objetivo del algoritmo es **minimizar la suma de distancias cuadradas** de cada punto a su centroide:
$$
\text{Inertia} = \sum_{i=1}^{n} \min_{j} ||x_i - \mu_j||^2
$$
donde $\mu_j$ es el centroide del cluster ( j ).

df["Cluster"] = kmeans.labels_

* `kmeans.labels_` contiene un número entero (0 o 1 en este caso) indicando a qué grupo pertenece cada respuesta.

## 🧠 Código completo con visualización



```python
# ================================================
# 📚 Análisis de clusters en texto con TF-IDF y KMeans
# ================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd

# Supongamos que tienes tu DataFrame df con una columna 'Respuesta_limpia'
# Ejemplo mínimo:
# df = pd.DataFrame({
#     "Respuesta_limpia": [
#         "me gusta aprender con ejemplos",
#         "prefiero clases teóricas",
#         "las prácticas ayudan a entender",
#         "la teoría es fundamental para comprender",
#         "aprendo mejor haciendo ejercicios"
#     ]
# })

# 1️⃣ Vectorización TF-IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["Respuesta_limpia"])

# 2️⃣ Aplicar K-means con 2 clusters
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)
df["Cluster"] = kmeans.labels_

# 3️⃣ Reducir dimensiones a 2 componentes para visualización
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X.toarray())

# 4️⃣ Crear DataFrame con componentes principales
df_pca = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "Cluster": df["Cluster"]
})

# 5️⃣ Visualización de los clusters
plt.figure(figsize=(8, 6))
for cluster in df_pca["Cluster"].unique():
    subset = df_pca[df_pca["Cluster"] == cluster]
    plt.scatter(subset["PC1"], subset["PC2"], label=f"Cluster {cluster}", s=80, alpha=0.7)

plt.title("Visualización de clusters de respuestas (PCA + K-means)", fontsize=14)
plt.xlabel("Componente principal 1")
plt.ylabel("Componente principal 2")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

# 6️⃣ Visualizar los textos por cluster (para interpretación)
for c in sorted(df["Cluster"].unique()):
    print(f"\n🔹 Cluster {c}:")
    for r in df[df["Cluster"] == c]["Respuesta_limpia"]:
        print("  -", r)
```

---



## 🔍 Interpretación de la gráfica

* Cada punto representa una **respuesta**.
* Los colores muestran los **grupos semánticos** detectados.
* Si los puntos de un color están cercanos entre sí y alejados de los otros, significa que **K-means ha identificado temas distintos**.

Por ejemplo:

* Un grupo podría corresponder a respuestas más **prácticas o aplicadas** (“me gusta aprender haciendo”).
* Otro grupo podría reflejar respuestas más **teóricas o reflexivas** (“la teoría es fundamental”).
