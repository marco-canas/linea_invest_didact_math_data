##### **Actividad 1: Organización de datos etnográficos**  
- Crear un archivo JSON/CSV con datos simulados de entrevistas: 

````python
import pandas as pd
  
datos_etnograficos = [
      {"id": 1, "municipio": "Caucasia", "entrevistado": "Líder comunitario", "texto": "El río ya no da lo mismo...", "tema": "agua"},
      {"id": 2, "municipio": "El Bagre", "entrevistado": "Agricultor", "texto": "Los cultivos de coca dejaron tierras inservibles...", "tema": "tierra"},
  ]
  
df = pd.DataFrame(datos_etnograficos)
df.to_csv("entrevistas_bajo_cauca.csv", index=False)
  
````


##### **Actividad 2: Análisis de contenido básico**  
- Contar frecuencia de palabras clave con `NLTK`:  
  ```python
  from nltk.tokenize import word_tokenize
  from collections import Counter
  
  texto = " ".join(df["texto"].tolist())
  palabras = word_tokenize(texto.lower(), language="spanish")
  filtradas = [p for p in palabras if len(p) > 4 and p.isalpha()]  # Eliminar palabras cortas
  Counter(filtradas).most_common(5)  # Top 5 palabras recurrentes
  ```

##### **Actividad 3: Visualización de temas**  
- Gráfico de barras con temas predominantes:  
  ```python
  df["tema"].value_counts().plot(kind="bar", color="green")
  plt.title("Temas más abordados en entrevistas")
  plt.xlabel("Tema")
  plt.ylabel("Frecuencia")
  ```
