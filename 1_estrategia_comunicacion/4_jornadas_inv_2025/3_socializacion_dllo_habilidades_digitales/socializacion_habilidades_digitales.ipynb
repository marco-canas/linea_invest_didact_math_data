# Socialización Proyecto Investigativo en Desarrollo de habilidades digitales  



## Desarrollo de habilidades digitales en los profesionales de administración de empresas: caso G-líderes UdeA Caucasia

# Introducción  

1. Contexto y Justificación  
   
En Colombia, las cifras de la UNESCO (2021) y el Ministerio de Educación Nacional (MEN, 2023) evidencian una amplia brecha en habilidades digitales: solo el 34,7% de jóvenes y adultos poseen competencias básicas, el 26,4% intermedias y el 4,6% avanzadas, cifras inferiores al promedio de los países de la OCDE. Esta situación se acentúa en las regiones, donde el Índice de Brecha Digital Regional alcanzó el 56% en 2021.


```python
# ============================================================
# Diagrama circular sobre brecha en habilidades digitales
# Fuente: UNESCO (2021), MEN (2023)
# ============================================================

import matplotlib.pyplot as plt

# Datos
categorias = ['Competencias básicas', 'Competencias intermedias', \
    'Competencias avanzadas']
porcentajes = [34.7, 26.4, 4.6]

# Crear figura
plt.figure(figsize=(7, 7))
plt.pie(
    porcentajes,
    labels=categorias,
    autopct='%1.1f%%',
    startangle=90,
    colors=['#4e79a7', '#f28e2b', '#e15759'],
    wedgeprops={'edgecolor': 'white'}
)

# Título
plt.title('''Distribución de habilidades digitales en Colombia\nUNESCO (2021) 
          y MEN (2023)''', fontsize=12, fontweight='bold')

# Agregar texto adicional sobre la brecha regional
plt.text(-1.3, -1.2,
         'Índice de Brecha Digital Regional: 56% (2021)\n'
         'Por debajo del promedio OCDE',
         fontsize=10,
         ha='left',
         va='center',
         bbox=dict(facecolor='whitesmoke', alpha=0.8, edgecolor='gray'))

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Diagrama circular interactivo sobre habilidades digitales
# Fuente: UNESCO (2021), MEN (2023)
# ============================================================

import plotly.graph_objects as go

# Datos
categorias = ['Competencias básicas', 'Competencias intermedias', \
    'Competencias avanzadas', 'no tiene competencias']
porcentajes = [34.7, 26.4, 4.6, 100-(34.7+ 26.4+ 4.6)]

# Crear figura interactiva
fig = go.Figure(
    data=[go.Pie(
        labels=categorias,
        values=porcentajes,
        textinfo='label+percent',
        insidetextorientation='radial',
        marker=dict(colors=['#4e79a7', '#f28e2b', '#e15759'], \
            line=dict(color='white', width=2)),
        hovertemplate='%{label}<br>%{value}%<extra></extra>'
    )]
)

# Personalización del diseño
fig.update_layout(
    title={
        'text': 'Distribución de habilidades digitales en Colombia<br><sup>UNESCO (2021) y MEN (2023)</sup>',
        'y':0.93,
        'x':0.5,
        'xanchor': 'center',
        'yanchor': 'top',
        'font': dict(size=18)
    },
    annotations=[
        dict(
            text="Índice de Brecha Digital Regional: 56% (2021)<br>Por debajo del promedio OCDE",
            x=0.5, y=-0.1,
            showarrow=False,
            font=dict(size=12, color='black'),
            align='center'
        )
    ],
    showlegend=False,
    height=600,
    width=600
)

fig.show()
```

# 2. Propósito de la Intervención  

El propósito general fue desarrollar habilidades digitales básicas, intermedias y avanzadas en los estudiantes del programa G-Líderes de la Universidad de Antioquia, sede Caucasia. 

La meta fue que los participantes adquirieran competencias que les permitieran usar tecnologías como Power BI y Python, y aplicar metodologías de análisis de datos y Machine Learning  y Deep Learning, a problemas administrativos reales.

# 3. Enfoque Teórico y Conceptual   
   
El proyecto se sustentó en:
*	La Teoría del Capital Humano (Becker, 1964): la formación en habilidades digitales es una inversión que mejora la productividad, la empleabilidad y la capacidad de innovación de los futuros profesionales.
*	La Transformación Digital (Gong & Ribiere, 2021): entendida como un cambio estratégico que redefine procesos, modelos de negocio y valor organizacional mediante el uso de tecnologías digitales.
*	El Ecosistema Digital: como entorno de aprendizaje colaborativo y tecnológico que potencia la innovación y la competitividad.



# 4. Metodología de la Intervención   
Se adoptó un diseño mixto con cuatro fases articuladas:  

1.	Diagnóstico: aplicación de encuestas y pruebas prácticas para identificar el nivel inicial de habilidades digitales.
2.	Intervención: desarrollo de talleres prácticos sobre Power BI, Python, análisis de datos, Machine Learning y Deep Learning aplicados a la gestión empresarial.
3.	Evaluación: medición de los avances mediante pruebas posteriores, encuestas de satisfacción y grupos focales.
4.	Análisis: procesamiento de la información mediante herramientas como pandas, Statsmodels, Seaborn y Scikit-learn para generar reportes y conclusiones con soporte estadístico y visual.


# 5. Avances y Logros
Hasta la finalización de la intervención se lograron los siguientes resultados:
*	Se han ejecutado nueve talleres previstos, centrados en la obtención, visualización tabular y gráfica de datos y en la construcción de modelos de machine learning y deep learning, de regresión y clasificación.
*	Los participantes mejoraron su comprensión y manejo de datos, adquirieron nociones sobre inteligencia de negocios y aprendieron a emplear herramientas digitales en contextos administrativos.
*	Se fortaleció una comunidad de práctica digital entre los estudiantes, que fomenta la autonomía, el aprendizaje colaborativo y la apropiación tecnológica.



# 6. Desafíos Identificados  

 A lo largo de lo que llevamos hasta ahora, identificamos los siguientes retos:  
*	Brecha digital entre los estudiantes, reflejada en desigualdad en acceso y manejo de tecnologías. El lenguaje Python es desconocido y su aplicación a la ciencia de datos. 
•	Limitaciones de infraestructura tecnológica del campus. El internet falla mucho.
•	Resistencia al cambio por parte de algunos actores académicos, como estudiantes de los programas académicos y muchos profesores.  
•	Falta de recursos y formación específica en herramientas digitales aplicadas a la administración.
•	Desarticulación parcial entre formación académica y las demandas del mercado laboral digital.



# 7. Conclusiones y Proyección    

* La intervención demostró que es posible reducir la brecha digital mediante estrategias formativas activas, basadas en el uso de tecnologías abiertas como VS Code, Cuadernos Jupyter, GitHub y YouTube y la pedagogía por proyectos.  
    
* Los resultados preliminares confirman un aumento en la apropiación tecnológica, el pensamiento analítico, y la motivación hacia la innovación digital entre los participantes.    
  
* En su siguiente fase, el equipo investigador continuará con la intervención en deep learning y visualización de datos con matplotlib  plotly y la medición final de impacto, buscando diseñar un modelo replicable de formación digital para estudiantes de Administración en regiones con condiciones similares a Caucasia.
