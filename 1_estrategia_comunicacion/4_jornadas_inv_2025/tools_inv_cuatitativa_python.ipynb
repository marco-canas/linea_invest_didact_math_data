# **Master Class: Herramientas de Python para Investigación Cualitativa**  
**Dirigido a:** Investigadores del Bajo Cauca Antioqueño (Universidad de Antioquia, Campus Caucasia)  
**Duración:** 4 horas (presencial o híbrida)  
**Nivel:** Básico a intermedio  

---



## **Objetivo General**  
Capacitar a investigadores en el uso de herramientas de Python para el análisis y procesamiento de datos cualitativos, optimizando sus procesos de investigación en áreas como ciencias sociales, humanidades y estudios regionales.  

---



## **Estructura del Taller**  

### **1. Introducción (30 min)**  
- Presentación del taller y expectativas.  
- Importancia de Python en investigación cualitativa.  
- Casos de aplicación en estudios regionales (ej: análisis de discurso, entrevistas, etnografías).  
- Instalación de entornos recomendados (Google Colab, Jupyter Notebook, Anaconda).  


### **1. Introducción (30 min)**  

#### **Presentación del taller y expectativas (5 min)**  
- **Bienvenida:** Saludo inicial, presentación del facilitador y objetivos de la sesión.  
- **Dinámica rompehielos:** Pregunta rápida a los participantes: *"¿Qué esperan aprender hoy sobre Python y su aplicación en investigación cualitativa?"* (Respuestas breves en chat o verbalmente).  
- **Ruta del taller:** Explicación breve de los módulos y metodología (teoría + práctica).  

#### **Importancia de Python en investigación cualitativa (10 min)**  
- **Contexto global:**  
  - Cómo herramientas digitales están transformando la investigación social (ej: minería de textos, automatización de análisis).  
  - Ventajas de Python: **gratuito, flexible, con librerías especializadas** vs. software tradicional (NVivo, Atlas.ti).  
- **Enfoque local:**  
  - Relevancia para el Bajo Cauca: *"¿Cómo puede ayudar Python a estudiar problemáticas como conflicto, minería o dinámicas culturales con datos cualitativos?"*.  
  - Ejemplo concreto: *"Automatizar el análisis de 100 entrevistas sobre desplazamiento en la región"*.  



#### **Casos de aplicación en estudios regionales (10 min)**  
- **Proyectos reales o simulados:**  
  1. **Análisis de discurso:** Extraer patrones en declaraciones de líderes sociales o medios locales.  
  2. **Etnografías digitales:** Procesar diarios de campo o notas de observación con NLP.  
  3. **Entrevistas:** Codificación automática de temas recurrentes (ej: "violencia", "resistencia").  
- **Demo rápida (opcional):** Mostrar un gráfico de *wordcloud* generado con datos de un taller comunitario en Caucasia.  



#### **Instalación de entornos recomendados (5 min)**  
- **Opciones accesibles:**  
  - **Google Colab** (recomendado para principiantes, sin instalación).  
  - **Jupyter Notebook** (guía rápida de instalación con Anaconda).  
- **Enlace a repositorio:** Compartir carpeta con:  
  - Instrucciones paso a paso.  
  - Dataset de ejemplo (ej: transcripción de una entrevista ficticia sobre cultivos en el Bajo Cauca).  
  - Cuaderno de Jupyter preconfigurado.  

---



#### **Notas para el facilitador:**  
- **Adaptar el lenguaje:** Evitar tecnicismos excesivos; vincular siempre los ejemplos al contexto local.  
- **Inclusión:** Si hay participantes con cero experiencia en programación, recalcar que se partirá desde lo básico.

### **2. Fundamentos de Python para Investigación (1 hora)**  
- Sintaxis básica (variables, listas, diccionarios).  
- Manejo de archivos (lectura/escritura de textos, CSV, JSON).  
- Librerías clave:  
  - `Pandas` (estructuración de datos cualitativos).  
  - `NLTK`/`spaCy` (procesamiento de lenguaje natural).  
  - `Matplotlib`/`WordCloud` (visualización de datos textuales).  



### **3. Análisis Cualitativo con Python (1.5 horas)**  
- **Preprocesamiento de textos:**  
  - Tokenización, limpieza de datos (stopwords, lematización).  
  - Ejemplo práctico: Análisis de entrevistas o discursos políticos del Bajo Cauca.  
- **Técnicas de análisis:**  
  - Frecuencia de términos.  
  - Análisis de sentimientos con `TextBlob`.  
  - Codificación automatizada (ej: identificación de categorías temáticas).  

### **4. Visualización y Presentación de Resultados (1 hora)**  
- Gráficos de barras para frecuencias de palabras.  
- Nubes de palabras (`WordCloud`).  
- Mapas conceptuales simples con `NetworkX`.  
- Exportación de resultados (informes en PDF, HTML o interactivos).  

### **5. Cierre y Discusión (30 min)**  
- Retroalimentación y casos de aplicación en proyectos reales.  
- Recomendaciones de recursos (cursos, libros, repositorios).  
- Red de colaboración entre investigadores.
