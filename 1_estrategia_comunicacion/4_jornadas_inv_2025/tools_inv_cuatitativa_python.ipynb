#### **1. Preprocesamiento de Textos (40 min)**  
**Objetivo:** Preparar datos textuales (entrevistas, discursos, informes) para análisis automatizado.  



**A. Tokenización y Limpieza Básica**  
- **Ejemplo con entrevistas sobre minería ilegal:**  


````python

from nltk.tokenize import word_tokenize
import nltk
texto = "La minería ilegal en el Bajo Cauca contamina los ríos y afecta a las comunidades"
tokens = word_tokenize(texto.lower())  # Convertir a minúsculas y dividir en palabras
print(tokens)  # Resultado: ['la', 'minería', 'ilegal', 'en', 'el', 'bajo', 'cauca', ...]
nltk.download('punkt')
````

  

**B. Eliminación de Stopwords (palabras irrelevantes)**

````python
  
from nltk.corpus import stopwords
stopwords_es = stopwords.words('spanish') + ["bajo", "cauca"]  # Añadir términos locales no relevantes
tokens_limpios = [word for word in tokens if word not in stopwords_es]
print(tokens_limpios)  # Resultado: ['minería', 'ilegal', 'contamina', 'ríos', 'afecta', 'comunidades']
````

  

**C. Lematización (reducir palabras a su raíz)**  


````python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer('spanish')
lemas = [stemmer.stem(word) for word in tokens_limpios]
print(lemas)  # Resultado: ['miner', 'ilegal', 'contamin', 'ríos', 'afect', 'comun']
````

#### **2. Técnicas de Análisis (50 min)**  



**A. Frecuencia de Términos**  
- **Identificar temas clave en proyectos de sostenibilidad:**  
  ```python
  from collections import Counter
  frecuencias = Counter(tokens_limpios)
  print(frecuencias.most_common(5))  # Ejemplo: [('minería', 12), ('agua', 8), ('violencia', 6)]
  ```  



**B. Análisis de Sentimientos con `TextBlob`**  
- **Evaluar tono en noticias o discursos:**  
  ```python
  from textblob import TextBlob
  texto = "La deforestación en el Bajo Cauca es una tragedia ambiental"
  analisis = TextBlob(texto).translate(to='en')  # TextBlob funciona mejor en inglés
  print(analisis.sentiment.polarity)  # Rango: -1 (negativo) a 1 (positivo)
  ```

  
  - *Nota:* Traducir al inglés puede afectar precisión. Alternativa: Usar librerías en español como `pysentimiento`.  



**C. Codificación Automatizada (Identificación de Categorías Temáticas)**  
- **Ejemplo con entrevistas sobre economía local:**  
  ```python
  categorias = {"medio_ambiente": ["río", "contaminación", "deforestación"],
                "economía": ["cultivos", "ingresos", "minería"]}
  
  texto_entrevista = "Los cultivos de cacao son afectados por la contaminación del río"
  palabras = texto_entrevista.lower().split()
  
  temas = []
  for categoria, palabras_clave in categorias.items():
      if any(palabra in palabras for palabra in palabras_clave):
          temas.append(categoria)
  print(temas)  # Resultado: ['medio_ambiente', 'economía']
  ```

#### **3. Mapas Conceptuales con `NetworkX` (15 min)**  
**Objetivo:** Mapear relaciones entre actores y temas en proyectos territoriales.  

- **Ejemplo: Red de actores del desarrollo:**  
  ```python
  import networkx as nx
  import matplotlib.pyplot as plt

  G = nx.Graph()
  G.add_nodes_from(["UAO", "Gobernación", "JAC", "Mineros"])  # Actores clave
  G.add_edges_from([
      ("UAO", "Gobernación", {"weight": 3}),  # Cooperación fuerte
      ("UAO", "JAC", {"weight": 5}),         # Trabajo directo con comunidades
      ("JAC", "Mineros", {"weight": 1})      # Interacción débil
  ])

  pos = nx.spring_layout(G)
  nx.draw(G, pos, with_labels=True, node_color='#4F7942', font_weight='bold',
          node_size=2000, edge_color='gray', width=[d['weight'] for _, _, d in G.edges(data=True)])
  plt.title("Red de actores en proyectos de desarrollo sostenible", pad=20)
  plt.show()
  ```  
  - **Aplicación:**  
    - Visualizar alianzas institucionales.  
    - Mapear conflictos entre actores (ej: comunidades vs. empresas mineras).
