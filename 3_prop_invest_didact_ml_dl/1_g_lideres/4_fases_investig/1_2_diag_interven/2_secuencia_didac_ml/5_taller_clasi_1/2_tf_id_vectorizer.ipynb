# El módulo **`TfidfVectorizer`** de `sklearn.feature_extraction.text` 

es una herramienta clave en el procesamiento de lenguaje natural (NLP) para chatbots, especialmente en el ámbito de la administración de empresas. 

Su función principal es convertir textos (como preguntas de usuarios o respuestas del chatbot) en **vectores numéricos** basados en la métrica **TF-IDF** (*Term Frequency-Inverse Document Frequency*), que ayuda a cuantificar la importancia de cada palabra en un documento respecto a un conjunto de documentos.





### **Ejemplo aplicado a un chatbot de administración de empresas**
Imagina un chatbot que responde consultas sobre gestión empresarial, como:
- *"¿Cómo calcular el punto de equilibrio?"*
- *"¿Qué es un KPI?"*
- *"Métodos para evaluar la rentabilidad de un proyecto."*



#### **Paso 1: Preprocesamiento de datos**
Antes de usar `TfidfVectorizer`, se recomienda:
- Tokenizar y limpiar textos (eliminar stopwords, signos de puntuación, etc.).
- Normalizar palabras (ej.: convertir a minúsculas, usar lematización).



#### **Paso 2: Configuración de TfidfVectorizer**


```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Corpus de documentos (preguntas/respuestas del chatbot)
corpus = [
    "Cómo calcular el punto de equilibrio en finanzas",
    "Los KPIs son indicadores clave de rendimiento",
    "Evaluación de rentabilidad con el método VAN o TIR",
    "Qué es un balance general en contabilidad"
]

# Inicializar el vectorizador
vectorizer = TfidfVectorizer(stop_words=['en', 'de', 'con', 'el', 'qué', 'son'])

# Ajustar y transformar el corpus
X = vectorizer.fit_transform(corpus)

print(X) 
```


#### **Paso 3: Resultado (TF-IDF Matrix)**
`X` será una matriz donde:
- **Filas**: Cada documento (pregunta/respuesta).
- **Columnas**: Términos (palabras) del vocabulario.
- **Valores**: Peso TF-IDF de cada término en cada documento.



**Ejemplo de vocabulario generado:**


```python

print(vectorizer.get_feature_names_out())
# Output: ['balance', 'calcular', 'contabilidad', 'equilibrio', 'evaluación', 'finanzas', 'general', 'indicadores', 'key', 'kpis', 'método', 'punto', 'rendimiento', 'rentabilidad', 'van', 'tir']

```


#### **Paso 4: Uso en el chatbot**
1. **Entrenamiento**: 
   - Vectorizar preguntas frecuentes (FAQ) para crear una base de conocimiento.
2. **Inferencia**:
   - Cuando un usuario escriba una pregunta (ej.: *"¿Cómo evaluar la rentabilidad?"*), el chatbot vectoriza el texto y compara su similitud (usando **cosine similarity**) con las preguntas almacenadas para encontrar la respuesta más relevante.


```python

from sklearn.metrics.pairwise import cosine_similarity

# Nueva consulta del usuario
consulta = "métodos para evaluar rentabilidad"
consulta_vector = vectorizer.transform([consulta])

# Calcular similitud con el corpus
similitudes = cosine_similarity(consulta_vector, X)
respuesta_idx = similitudes.argmax()  # Índice de la respuesta más similar
print(corpus[respuesta_idx])
# Output: "Evaluación de rentabilidad con el método VAN o TIR"

```




### **¿Por qué es útil en administración de empresas?**
1. **Precisión en respuestas**: 
   - TF-IDF prioriza términos técnicos relevantes (ej.: "VAN", "KPI", "punto de equilibrio") sobre palabras genéricas.
2. **Eficiencia**:
   - Automatiza la clasificación de consultas sin necesidad de reglas manuales complejas.
3. **Escalabilidad**:
   - Fácil de actualizar añadiendo nuevos documentos al corpus.





### **Limitaciones a considerar**
- No captura contexto semántico (para eso se usan modelos como **Word2Vec** o **BERT**).
- Requiere un corpus amplio y representativo para resultados óptimos.

Si tu chatbot maneja jerga empresarial específica, puedes mejorar `TfidfVectorizer` añadiendo **n-grams** (`ngram_range=(1,2)`) para detectar frases clave como *"punto de equilibrio"* o *"flujo de caja"*.

# El ejemplo con un **dataset simulado de preguntas reales en administración de empresas** y un flujo completo para entrenar un chatbot básico usando `TfidfVectorizer`. 

Incluiré:

1. **Dataset de preguntas y respuestas** (FAQ de administración).
2. **Preprocesamiento** con limpieza de texto.
3. **Entrenamiento del vectorizador** y cálculo de similitud.
4. **Ejemplo de interacción con el chatbot**.





### **Paso 1: Dataset de Ejemplo (FAQ)**
Vamos a simular un pequeño dataset de preguntas frecuentes en administración:


```python

import pandas as pd

data = {
    "pregunta": [
        "¿Cómo se calcula el punto de equilibrio?",
        "¿Qué es un KPI en gestión empresarial?",
        "¿Cuál es la diferencia entre balance general y estado de resultados?",
        "¿Cómo evaluar la rentabilidad de un proyecto con VAN?",
        "¿Qué estrategias hay para reducir costos fijos?",
        "¿Para qué sirve un análisis SWOT en una empresa?"
    ],
    "respuesta": [
        "El punto de equilibrio se calcula dividiendo los costos fijos entre el margen de contribución por unidad.",
        "Un KPI (Key Performance Indicator) es una métrica para medir el rendimiento de procesos críticos, como ventas o productividad.",
        "El balance general muestra activos, pasivos y patrimonio en un momento dado, mientras que el estado de resultados muestra ingresos y gastos en un período.",
        "El Valor Actual Neto (VAN) descuenta flujos de caja futuros a valor presente. Si VAN > 0, el proyecto es rentable.",
        "Ejemplos: renegociar contratos de arrendamiento, automatizar procesos o subcontratar servicios no esenciales.",
        "El análisis SWOT identifica Fortalezas, Debilidades, Oportunidades y Amenazas para la toma de decisiones estratégicas."
    ]
}

df = pd.DataFrame(data)

```


---

### **Paso 2: Preprocesamiento de Texto**
Limpiamos las preguntas (eliminamos stopwords, acentos, etc.):


```python

import re
from sklearn.feature_extraction.text import TfidfVectorizer

def limpiar_texto(texto):
    # Convertir a minúsculas y eliminar acentos (simplificado)
    texto = texto.lower()
    texto = re.sub(r'[áéíóú]', lambda x: {'á':'a', 'é':'e', 'í':'i', 'ó':'o', 'ú':'u'}[x.group()],\
        texto)
    # Eliminar signos de puntuación
    texto = re.sub(r'[^\w\s]', '', texto)
    return texto

df['pregunta_limpia'] = df['pregunta'].apply(limpiar_texto)

```




### **Paso 3: Entrenar TfidfVectorizer**
Configuramos el vectorizador con parámetros para jerga empresarial:


```python

vectorizer = TfidfVectorizer(
    stop_words=['como', 'qué', 'para', 'el', 'en', 'la', 'los', 'se'],  # Stopwords personalizadas
    ngram_range=(1, 2),  # Capturar frases como "punto de equilibrio"
    max_features=1000     # Límite de términos para eficiencia
)

# Vectorizar las preguntas
X = vectorizer.fit_transform(df['pregunta_limpia'])

```


---

### **Paso 4: Función para Responder Preguntas**
Usamos **cosine similarity** para encontrar la pregunta más similar:


```python

from sklearn.metrics.pairwise import cosine_similarity

def responder_chatbot(consulta_usuario, df, vectorizer, X):
    # Limpiar consulta
    consulta_limpia = limpiar_texto(consulta_usuario)
    # Vectorizar consulta
    consulta_vector = vectorizer.transform([consulta_limpia])
    # Calcular similitud
    similitudes = cosine_similarity(consulta_vector, X)
    # Encontrar la respuesta más similar
    idx_max = similitudes.argmax()
    return df.iloc[idx_max]['respuesta']

```


---

### **Paso 5: Probemos el Chatbot**
Interacciones de ejemplo:


```python

# Ejemplo 1: Consulta sobre KPIs
consulta = "¿Qué son los indicadores clave de rendimiento?"
respuesta = responder_chatbot(consulta, df, vectorizer, X)
print(f"Chatbot: {respuesta}")
# Output: "Un KPI (Key Performance Indicator) es una métrica para medir el rendimiento de procesos críticos..."
```

```python

# Ejemplo 2: Consulta sobre rentabilidad
consulta = "¿Cómo saber si un proyecto es rentable?"
respuesta = responder_chatbot(consulta, df, vectorizer, X)
print(f"Chatbot: {respuesta}")
# Output: "El Valor Actual Neto (VAN) descuenta flujos de caja futuros a valor presente. Si VAN > 0, el proyecto es rentable."

```


---

### **Análisis de lo que Ocurre Detrás**
1. **TF-IDF** asigna pesos altos a términos técnicos como:
   - `"kpi"`, `"van"`, `"swot"` (importantes en administración).
   - Frases como `"punto de equilibrio"` (gracias a `ngram_range=(1,2)`).
2. **Stopwords personalizadas** evitan ruido en términos genéricos (`"qué"`, `"para"`).

---

### **Mejoras Opcionales**
1. **Añadir sinónimos**: 
   - Mapear "utilidad" → "rentabilidad" o "ganancia".
2. **Usar Stemming/Lematización**:
   - `"calcula"` y `"calcular"` se tratan como la misma raíz.
3. **Dataset más grande**:
   - Más preguntas para cubrir casos como `"cómo hacer un presupuesto"`.


```python

# Ejemplo de lematización con spaCy
import spacy
nlp = spacy.load("es_core_news_sm")

def lematizar(texto):
    doc = nlp(texto)
    return " ".join([token.lemma_ for token in doc])

df['pregunta_lematizada'] = df['pregunta_limpia'].apply(lematizar)

```


---

### **Conclusión**
Este flujo es ideal para chatbots de **FAQ empresariales** donde las preguntas son técnicas y los términos específicos (como `VAN`, `SWOT`, `KPI`) son cruciales. Para consultas más complejas (ej.: "Explícame el VAN como si fuera un niño de 5 años"), modelos como **BERT** serían mejores.
