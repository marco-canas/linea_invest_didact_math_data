# El módulo python-docx   


La sentencia `from docx import Document` es una línea de código en Python que se utiliza para importar la clase `Document` del módulo `python-docx`, una biblioteca popular para trabajar con archivos de Microsoft Word (extensión `.docx`).



### Explicación detallada:


1. **`from docx`**:  
   - `docx` es el nombre del módulo (paquete) que contiene las herramientas para manipular archivos Word (.docx).  
   - Este módulo no forma parte de la biblioteca estándar de Python, por lo que debe instalarse previamente con:  
     ```bash
     pip install python-docx
     ```



2. **`import Document`**:  
   - `Document` es la clase principal del módulo `docx`. Representa un documento de Word y permite:  
     - Crear un nuevo documento.  
     - Abrir y modificar documentos existentes.  
     - Añadir párrafos, tablas, imágenes, etc.  



### Ejemplo de uso:


````python

from docx import Document

# Crear un nuevo documento
doc = Document()

# Añadir un título y un párrafo
doc.add_heading('Título del documento', level=1)
doc.add_paragraph('Este es un párrafo de ejemplo.')

# Guardar el documento
doc.save('ejemplo.docx')

````


### Funcionalidades clave de `Document`:
- **Manipulación de texto**: Añadir párrafos, estilos (negrita, cursiva), etc.  
- **Tablas**: Crear tablas con `add_table()`.  
- **Formato**: Modificar fuentes, alineación, márgenes.  
- **Encabezados/pies de página**: Usar `sections` para personalizarlos.  

### Notas importantes:
- El módulo `python-docx` solo funciona con archivos `.docx` (no con el antiguo formato `.doc`).  
- Para leer texto existente de un documento, se usan métodos como `paragraphs` o `tables` sobre el objeto `Document`.
