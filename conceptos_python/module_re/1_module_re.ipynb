# El Módulo `re` (Expresiones Regulares)   

El módulo **`re`** en Python es una herramienta fundamental para trabajar con **expresiones regulares** (*regular expressions* o *regex*). 

Permite realizar búsquedas, manipulaciones y validaciones de texto basadas en patrones complejos. 

Es parte de la biblioteca estándar de Python, por lo que no requiere instalación adicional.





## **Funcionalidades principales del módulo `re`**


### 1. **Búsqueda de patrones**


   - **`re.search()`**: Busca un patrón en un string y devuelve la primera coincidencia (como objeto `Match`).


````python

import re  # importa el módulo para expresiones regulares en string  
texto = "Python es genial" # texto a buscar la palabra
resultado = re.search(r"genial", texto) # busca la palabra en el texto
print(resultado.group())  # Salida: "Python"

````

   - **`re.match()`**: Similar a `search()`, pero solo busca al **inicio** del string.
     ```python


````python
resultado = re.match(r"Python", texto)  # Coincide
resultado 
````

````python
texto = 'es maravilloso python'
resultado = re.match(r"es", texto)     # No coincide (no está al inicio)
resultado 
````


### 2. **Búsqueda global**
   - **`re.findall()`**: Devuelve **todas** las coincidencias como una lista.


````python
texto = "10 manzanas, 5 peras, 3 plátanos"
numeros = re.findall(r"\d+", texto)
print(numeros)  # Salida: ['10', '5', '3']
     
````

   - **`re.finditer()`**: Similar a `findall()`, pero devuelve un iterador de objetos `Match`.



### 3. **Sustitución de texto**
   - **`re.sub()`**: Reemplaza todas las coincidencias de un patrón con otro texto.


````python
texto = "Hola mundo"
nuevo_texto = re.sub(r"mundo", "Python", texto)
print(nuevo_texto)  # Salida: "Hola Python"
     
````


### 4. **División de strings**


   - **`re.split()`**: Divide un string usando un patrón como delimitador.


````python
texto = "Python,Java;C++|Rust"
palabras = re.split(r"[,;|]", texto)
print(palabras)  # Salida: ['Python', 'Java', 'C++', 'Rust']

````


---

## **Sintaxis básica de expresiones regulares**


| **Patrón**  | **Descripción**                                                                 |
|-------------|-------------------------------------------------------------------------------|
| `\d`        | Coincide con un dígito (0-9).                                                |
| `\w`        | Coincide con un carácter alfanumérico (letras, números o `_`).               |
| `\s`        | Coincide con un espacio en blanco (espacio, tab, salto de línea).            |
| `.`         | Coincide con cualquier carácter (excepto salto de línea).                    |
| `*`         | Coincide con 0 o más repeticiones del patrón anterior.                       |
| `+`         | Coincide con 1 o más repeticiones del patrón anterior.                       |
| `{n}`       | Coincide exactamente con `n` repeticiones del patrón anterior.               |
| `[a-z]`     | Coincide con cualquier letra minúscula.                                      |
| `^`         | Coincide al inicio del string.                                               |
| `$`         | Coincide al final del string.                                                |

---



## **Ejemplo avanzado: Validar un email**


````python

import re

def validar_email(email):
    patron = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    return re.match(patron, email) is not None

print(validar_email("usuario@example.com"))  # True
````

````python
print(validar_email("usuario@.com"))         # False

````

## **Ventajas del módulo `re`**
- **Eficiencia**: Optimizado para operaciones complejas con texto.
- **Flexibilidad**: Permite definir patrones personalizados para casi cualquier necesidad.
- **Integración**: Funciona perfectamente con otras bibliotecas de Python (ej: procesamiento de logs, web scraping).
